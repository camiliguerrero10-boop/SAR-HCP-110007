```python
# =============================================================================
# MODELO SAR – SUJETO HCP 110007
# Versión pública (descarga automática desde Drive)
# =============================================================================

# %% CELDA 0 – Instalación
%pip install -q gdown nilearn plotly libpysal spreg scikit-learn pandas scipy nibabel

# %% CELDA 1 – Librerías
import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd
from scipy.spatial.distance import cdist
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
import plotly.graph_objects as go
from IPython.display import display
from nilearn import datasets, surface
from libpysal.weights import W
from esda.moran import Moran
import nibabel as nib
import gdown

SUBJECT_ID = "110007"
print(f"Trabajando con el sujeto HCP: {SUBJECT_ID}")

# %% CELDA 2 – Descarga automática del archivo desde Drive público
print("\nDescargando dtseries desde Google Drive (puede tardar varios minutos)...")

os.makedirs("datos_hcp", exist_ok=True)

# ID del archivo público
DTSERIES_ID = "10M1JMHYqW7d46ANe5e4xFGjBZuUC5G5t"

path_dtseries = "datos_hcp/rfMRI_REST.dtseries.nii"

if not os.path.exists(path_dtseries):
    gdown.download(id=DTSERIES_ID, output=path_dtseries, quiet=False)
else:
    print("Archivo ya existe, no se vuelve a descargar.")

path_thickness = None   # No hay thickness en este caso

print("→ dtseries listo:", path_dtseries)

# %% CELDA 3 – Geometría cortical (fsaverage5 + Destrieux)
print("\nCargando superficie cortical real y atlas Destrieux...")
fsavg = datasets.fetch_surf_fsaverage(mesh="fsaverage5")
atlas = datasets.fetch_atlas_surf_destrieux()

nodes = []
coords_list = []

for hemi in ["left", "right"]:
    mesh = surface.load_surf_mesh(fsavg[f"pial_{hemi}"])
    coords_hemi = np.asarray(mesh.coordinates)
    labels = np.asarray(atlas[f"map_{hemi}"]).ravel()

    for parcel_id in sorted(np.unique(labels)):
        if parcel_id == 0:
            continue
        idx = np.where(labels == parcel_id)[0]
        if len(idx) < 8:
            continue
        centroid = coords_hemi[idx].mean(axis=0)
        nodes.append({
            "node": len(nodes),
            "subject": SUBJECT_ID,
            "hemi": hemi,
            "parcel_id": int(parcel_id),
            "x": float(centroid[0]),
            "y": float(centroid[1]),
            "z": float(centroid[2]),
            "n_vertices": len(idx)
        })
        coords_list.append(centroid)

nodes_df = pd.DataFrame(nodes)
coords   = np.array(coords_list)
N        = len(nodes_df)
print(f"Número de regiones corticales: {N}")
display(nodes_df.head())

# %% CELDA 4 – Carga de datos
print("\n" + "="*70)
print("CARGA DE DATOS")
print("="*70)

USE_REAL_FC = False
USE_REAL_THICKNESS = False

# Functional Connectivity
print("Cargando dtseries...")
cifti = nib.load(path_dtseries)
data  = cifti.get_fdata()
print(f"  Shape dtseries: {data.shape}")

n_cort = min(59412, data.shape[1])
ts_cort = data[:, :n_cort].T

step = n_cort // N
ts_parcel = np.array([ts_cort[i*step:(i+1)*step].mean(axis=0) for i in range(N)])
C_func = np.corrcoef(ts_parcel)
np.fill_diagonal(C_func, 0.0)
USE_REAL_FC = True
print("  → Matriz de conectividad funcional REAL extraída")

# Grosor sintético (no hay archivo de thickness)
np.random.seed(42)
D = cdist(coords, coords)
np.fill_diagonal(D, np.inf)

K = np.exp(-D / 20)
thickness = 2.5 + 0.35 * np.random.multivariate_normal(np.zeros(N), K + 0.12*np.eye(N))
thickness = np.clip(thickness, 1.8, 3.3)

fc_strength = np.abs(C_func).sum(axis=1)
fc_strength = (fc_strength - fc_strength.mean()) / fc_strength.std()

# Variable dependiente sintética
X = np.column_stack([np.ones(N), thickness, fc_strength])
beta_true = np.array([1.1, 0.42, 0.28])
rho_true  = 0.36

W_true = 1.0 / (D + 1e-5)
W_true = W_true / W_true.sum(axis=1, keepdims=True)
np.fill_diagonal(W_true, 0.0)

y = np.linalg.solve(
    np.eye(N) - rho_true * W_true,
    X @ beta_true + np.random.normal(0, 0.4, N)
)

nodes_df["grosor"]      = thickness
nodes_df["fc_strength"] = fc_strength
nodes_df["DD"]          = y

print(f"\nResumen datos:")
print(f"  Grosor real     : {USE_REAL_THICKNESS}")
print(f"  FC real         : {USE_REAL_FC}")
print(f"  N regiones      : {N}")
display(nodes_df[["parcel_id","hemi","grosor","fc_strength","DD"]].describe().round(3))

# %% CELDA 5 – Matrices de pesos
def row_standardize(Wmat):
    Wmat = np.asarray(Wmat, dtype=float).copy()
    np.fill_diagonal(Wmat, 0.0)
    row_sum = Wmat.sum(axis=1, keepdims=True)
    return Wmat / np.maximum(row_sum, 1e-12)

W1 = row_standardize(1.0 / (D + 1e-5))

nn = NearestNeighbors(n_neighbors=7).fit(coords)
_, idx = nn.kneighbors(coords)
W2 = np.zeros((N, N))
for i in range(N):
    W2[i, idx[i, 1:]] = 1
W2 = row_standardize(np.maximum(W2, W2.T))

abs_c = np.abs(C_func)
thresh = np.percentile(abs_c[np.triu_indices(N, 1)], 85)
W3 = row_standardize(np.where(abs_c >= thresh, abs_c, 0.0))

from sklearn.neighbors import radius_neighbors_graph
W4_raw = radius_neighbors_graph(coords, radius=18, mode="connectivity").toarray()
W4 = row_standardize(np.maximum(W4_raw, W4_raw.T))

weights = {
    "W1_Distancia_Inversa": W1,
    "W2_kNN":               W2,
    "W3_Funcional":         W3,
    "W4_Contigüidad":       W4
}
print("Matrices de pesos construidas.")

# %% CELDA 6 – Estimación SAR
def sar_2sls(y, X, Wmat, name):
    Wy = Wmat @ y
    Z = np.column_stack([X, Wmat @ X[:, 1:]])
    gamma = np.linalg.lstsq(Z, Wy, rcond=None)[0]
    Wy_hat = Z @ gamma
    Xaug = np.column_stack([X, Wy_hat])
    b = np.linalg.lstsq(Xaug, y, rcond=None)[0]
    resid = y - Xaug @ b
    r2 = 1 - np.sum(resid**2) / np.sum((y - y.mean())**2)

    try:
        w_pysal = W.from_adjlist(
            pd.DataFrame({
                "focal": np.repeat(np.arange(N), N),
                "neighbor": np.tile(np.arange(N), N),
                "weight": Wmat.ravel()
            }).query("weight > 0 and focal != neighbor")
        )
        mi = Moran(resid, w_pysal, permutations=99)
        moran_i, moran_p = mi.I, mi.p_sim
    except:
        moran_i, moran_p = np.nan, np.nan

    return {
        "Matriz": name,
        "ρ": b[-1],
        "β_const": b[0],
        "β_grosor": b[1],
        "β_fc": b[2],
        "R²": r2,
        "Moran_I_resid": moran_i,
        "Moran_p": moran_p
    }

results = [sar_2sls(y, X, W, name) for name, W in weights.items()]
results_df = pd.DataFrame(results)
print("\nResultados del modelo SAR – Sujeto 110007")
display(results_df.round(4))

# %% CELDA 7 – Visualización 3D
print("\nGenerando grafo 3D...")

fig = go.Figure()

for hemi in ["left", "right"]:
    mesh = surface.load_surf_mesh(fsavg[f"pial_{hemi}"])
    xyz = np.asarray(mesh.coordinates)
    faces = np.asarray(mesh.faces)[::4]
    fig.add_trace(go.Mesh3d(
        x=xyz[:,0], y=xyz[:,1], z=xyz[:,2],
        i=faces[:,0], j=faces[:,1], k=faces[:,2],
        color="#A8B4C4", opacity=0.14,
        name=f"Superficie {hemi}", hoverinfo="skip",
        showlegend=True
    ))

fig.add_trace(go.Scatter3d(
    x=coords[:,0], y=coords[:,1], z=coords[:,2],
    mode="markers",
    marker=dict(
        size=7,
        color=y,
        colorscale="RdYlBu_r",
        cmin=np.percentile(y, 5), cmax=np.percentile(y, 95),
        opacity=0.95,
        colorbar=dict(title="DD", x=1.02, thickness=15)
    ),
    text=[f"Parcela {r.parcel_id} ({r.hemi})<br>DD={r.DD:.2f}<br>Grosor={r.grosor:.2f}<br>FC={r.fc_strength:.2f}"
          for r in nodes_df.itertuples()],
    hovertemplate="%{text}<extra></extra>",
    name="Parcelas Destrieux"
))

def add_edges(Wmat, name, color, max_edges=70):
    iu, ju = np.triu_indices(N, 1)
    vals = Wmat[iu, ju]
    top = np.argsort(vals)[-min(max_edges, (vals > 0).sum()):]
    xx, yy, zz = [], [], []
    for a, b in zip(iu[top], ju[top]):
        xx += [coords[a,0], coords[b,0], None]
        yy += [coords[a,1], coords[b,1], None]
        zz += [coords[a,2], coords[b,2], None]
    fig.add_trace(go.Scatter3d(
        x=xx, y=yy, z=zz, mode="lines",
        line=dict(color=color, width=2.2),
        name=name, visible="legendonly"
    ))

add_edges(W1, "W1 · Distancia inversa", "rgba(70,130,220,0.55)")
add_edges(W2, "W2 · k-NN", "rgba(40,170,110,0.60)")
add_edges(W3, "W3 · Funcional", "rgba(220,60,50,0.70)")
add_edges(W4, "W4 · Contigüidad", "rgba(150,80,200,0.55)")

fig.update_layout(
    title=dict(
        text=f"Modelo SAR – Sujeto HCP {SUBJECT_ID}<br>"
             f"<sup>Geometría real fsaverage5 + Destrieux | FC real: {USE_REAL_FC} | Grosor real: {USE_REAL_THICKNESS}</sup>",
        x=0.5
    ),
    template="plotly_dark",
    width=1200, height=880,
    scene=dict(
        xaxis_visible=False, yaxis_visible=False, zaxis_visible=False,
        aspectmode="data",
        camera=dict(eye=dict(x=1.65, y=1.45, z=1.05))
    ),
    legend=dict(x=0.01, y=0.98, bgcolor="rgba(0,0,0,0.3)"),
    margin=dict(l=0, r=0, t=90, b=0)
)
fig.show()

print("\n✓ Pipeline completado.")
```
